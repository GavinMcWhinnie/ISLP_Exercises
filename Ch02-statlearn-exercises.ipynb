**Exercise 9**

This exercise uses the Auto data set, which can be found on https://www.statlearning.com/resources-python.

```python
# Load the Auto.csv file into a Dataframe
filename = 'Datasets/Auto.csv'
Auto = pd.read_csv(filename, na_values=['?'])
```

```python
# Make sure that the missing values have been removed
Auto = Auto.dropna()

# Set the first column with car names to be the index
Auto = Auto.set_index('name')
```

```python
Auto
```

(a) Quantitative: mpg, cylinders, displacement, horsepower, weight, acceleration, year
Qualitative: origin

```python
# Work out the range of each quantitative predictor
quantative_predictors = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'year']

for predictor in quantative_predictors:
    print('For the predictor \'{0}\', the range is:'.format(predictor))
    pred_min = np.min(Auto[predictor])
    pred_max = np.max(Auto[predictor])
    pred_range = pred_max - pred_min
    print(f"Range: {pred_range}.")
```

```python
# Work out the mean and standard deviation 
for predictor in quantative_predictors:
    print('For the predictor \'{0}\', the mean and standard deviation are:'.format(predictor))
    pred_mean = np.mean(Auto[predictor])
    pred_std = np.std(Auto[predictor])
    print(f"Mean: {pred_mean}, Standard Deviation: {pred_std}.\n")
```

```python
# (d) Remove the 10th through 85th observations
Auto_dropped = Auto.drop(Auto.index[9:85])
```

```python
# Work out the mean and standard deviation for the remaining observations
for predictor in quantative_predictors:
    print('For the predictor \'{0}\', the mean and standard deviation of the remaining data are:'.format(predictor))
    pred_mean = np.mean(Auto_dropped[predictor])
    pred_std = np.std(Auto_dropped[predictor])
    print(f"Mean: {pred_mean}, Standard Deviation: {pred_std}.\n")
```

```python
pd.plotting.scatter_matrix(Auto[quantative_predictors], figsize=(12, 12))
```

We see that **displacement**, **horsepower**, **weight**, and **cylinders** show clear negative correlation with **mpg**, whilst **acceleration** and **year** show slight positive correlations.
